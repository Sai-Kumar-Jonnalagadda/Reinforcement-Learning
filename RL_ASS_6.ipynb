## Step 1: Environment Setup

First, we'll import the `gymnasium` library and create the specified environments.

```python
import gymnasium as gym

# Create the CliffWalking-v1 environment
env_cliffwalking = gym.make('CliffWalking-v1')
print('CliffWalking-v1 environment created.')
```

```python
# Create the Taxi-v4 environment
env_taxi = gym.make('Taxi-v4')
print('Taxi-v4 environment created.')
```

## Step 2: Implement the Random Agent

Now, let's create a `RandomAgent` function that takes an environment and a number of episodes as input. This function will simulate the agent's interaction by:

1.  Resetting the environment.
2.  Observing the initial state.
3.  Identifying available actions from the action space.
4.  Selecting an action randomly.
5.  Executing the action in the environment.
6.  Observing the next state and reward.
7.  Continuing until the episode terminates.
8.  Calculating and displaying the total reward and average reward obtained for each episode.

```python
import numpy as np

def run_random_agent(env, num_episodes):
    """
    Runs a random agent in the given environment for a specified number of episodes.

    Args:
        env: The Gymnasium environment to interact with.
        num_episodes: The number of episodes to run.

    Returns:
        A list of total rewards for each episode.
    """
    all_episode_rewards = []

    for episode in range(num_episodes):
        # 1. Reset the environment and observe the initial state.
        # The reset method returns an observation and an info dictionary.
        observation, info = env.reset()

        total_reward = 0
        terminated = False  # True when the episode ends due to reaching a terminal state
        truncated = False   # True when the episode ends due to a time limit or other truncation condition

        print(f"\n--- Episode {episode + 1}/{num_episodes} ---")

        while not terminated and not truncated:
            # 4. Identify the available actions using the environment's action space.
            # 5. Select an action randomly from the action space.
            action = env.action_space.sample()

            # 6. Execute the action in the environment.
            # 7. Observe the next state, reward, terminated, truncated, and info.
            observation, reward, terminated, truncated, info = env.step(action)

            total_reward += reward

        # 9. Calculate and display the total reward and average reward for the episode.
        print(f"Episode {episode + 1} finished. Total Reward: {total_reward}")
        all_episode_rewards.append(total_reward)

    env.close() # Close the environment once all episodes are done
    return all_episode_rewards

print("Random Agent function defined.")
```

## Step 3: Test the Random Agent

Now, let's test our `run_random_agent` function with both the `CliffWalking-v1` and `Taxi-v4` environments.

```python
num_episodes = 10

print("\n--- Running Random Agent on CliffWalking-v1 ---")
cliffwalking_rewards = run_random_agent(env_cliffwalking, num_episodes)
print(f"\nCliffWalking-v1 - Total rewards over {num_episodes} episodes: {cliffwalking_rewards}")
print(f"CliffWalking-v1 - Average reward per episode: {np.mean(cliffwalking_rewards):.2f}")
```

```python
print("\n--- Running Random Agent on Taxi-v4 ---")
taxi_rewards = run_random_agent(env_taxi, num_episodes)
print(f"\nTaxi-v4 - Total rewards over {num_episodes} episodes: {taxi_rewards}")
print(f"Taxi-v4 - Average reward per episode: {np.mean(taxi_rewards):.2f}")
```
