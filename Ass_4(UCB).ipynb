```python
import math

# ----------------------------------------------------
# Part A - Machine Initialization
# ----------------------------------------------------
machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10
c = 2  # exploration constant (standard UCB1 coefficient)

# ----------------------------------------------------4
# Part B - Initialize knowledge tables (N, R, Q, UCB)
# ----------------------------------------------------
N = [0] * M       # number of times each machine has been played
R = [0] * M       # cumulative reward per machine
Q = [0.0] * M     # estimated value (Q-value) per machine
UCB = [0.0] * M   # UCB value per machine

total_reward = 0

for t in range(1, N_play + 1):

    # ----------------------------------------------------
    # Part C - UCB Calculation & Machine Selection
    # ----------------------------------------------------
    unplayed = [i for i in range(M) if N[i] == 0]

    if unplayed:
        # Play every machine once first to avoid division by zero
        selected = unplayed[0]
        for i in range(M):
            UCB[i] = float('inf') if N[i] == 0 else \
                     Q[i] + c * math.sqrt(math.log(t) / N[i])
    else:
        for i in range(M):
            UCB[i] = Q[i] + c * math.sqrt(math.log(t) / N[i])
        selected = UCB.index(max(UCB))

    print(f"\nPlay {t}:")
    print("UCB values:", [round(u, 4) if u != float('inf') else 'inf' for u in UCB])
    print(f"--> Machine chosen: {machines[selected]}")

    # ----------------------------------------------------
    # Part D - Reward Collection
    # ----------------------------------------------------
    reward = float(input(f"Play with machine {machines[selected]}\nEnter reward: "))

    # Update N and R
    N[selected] += 1
    R[selected] += reward
    total_reward += reward

    # ----------------------------------------------------
    # Part E - Incremental Q-Value Update
    # Q(a) <- Q(a) + (1/N(a)) * (R - Q(a))
    # ----------------------------------------------------
    Q[selected] = Q[selected] + (1 / N[selected]) * (reward - Q[selected])

# ----------------------------------------------------
# Part F - Display Results
# ----------------------------------------------------
print("\n===================== FINAL RESULTS =====================")
print(f"{'Machine':<10}{'N (plays)':<12}{'R (total)':<12}{'Q (avg)':<10}")
for i in range(M):
    print(f"{machines[i]:<10}{N[i]:<12}{R[i]:<12}{round(Q[i], 4):<10}")

print(f"\nTotal reward received over {N_play} plays: {total_reward}")
```
