```python
# Creating the grid_env_det.py module
import os

module_content = """
class GridWorldDet:
    def __init__(self):
        # States: S0, S1, S2 (Terminal), S3, S4, S5 (Terminal)
        # Grid layout:
        # S0  S1  S2 (Terminal)
        # S3  S4  S5 (Terminal)
        self.states = ['S0', 'S1', 'S2', 'S3', 'S4', 'S5']
        self.terminal_states = ['S2', 'S5']

        # Transition dynamics: {state: {action: (next_state, reward)}}
        self.transitions = {
            'S0': {
                'Right': ('S1', -1),
                'Down': ('S3', -1),
                'Left': ('S0', -1),
                'Up': ('S0', -1)
            },
            'S1': {
                'Right': ('S2', 10), # reaching S2 gets +10
                'Down': ('S4', -1),
                'Left': ('S0', -1),
                'Up': ('S1', -1)
            },
            'S3': {
                'Right': ('S4', -1),
                'Down': ('S3', -1),
                'Left': ('S3', -1),
                'Up': ('S0', -1)
            },
            'S4': {
                'Right': ('S5', 5), # reaching S5 gets +5
                'Down': ('S4', -1),
                'Left': ('S3', -1),
                'Up': ('S1', -1)
            }
        }

    def step(self, state, action):
        if state in self.terminal_states:
            return state, 0
        return self.transitions[state][action]
"""

with open('grid_env_det.py', 'w') as f:
    f.write(module_content)

print("grid_env_det.py written successfully.")
```

```python
import grid_env_det

# Initialize environment
env = grid_env_det.GridWorldDet()

# Define policy
# Policy(П) = [S0: Right, S1: Down, S3: Right, S4: UP]
policy = {
    'S0': 'Right',
    'S1': 'Down',
    'S3': 'Right',
    'S4': 'Up'
}

gamma = 0.9 # Discount factor

def state_value_function(start_state, policy, env, gamma):
    """Calculates State Value by following the path until a terminal state is reached."""
    if start_state in env.terminal_states:
        return 0.0

    state = start_state
    discounted_return = 0.0
    discount = 1.0
    visited = [] # Detect infinite loops if policy doesn't reach terminal

    while state not in env.terminal_states:
        if state in visited and len(visited) > 100:
            # Prevent infinite loops in bad policies
            break
        visited.append(state)

        action = policy[state]
        next_state, reward = env.step(state, action)
        discounted_return += discount * reward
        discount *= gamma
        state = next_state

    return discounted_return

def action_value_function(start_state, action, policy, env, gamma):
    """Calculates Action Value by taking initial action, then following the policy."""
    if start_state in env.terminal_states:
        return 0.0

    # Take the first action
    next_state, reward = env.step(start_state, action)
    discounted_return = reward

    # Then follow the policy
    state = next_state
    discount = gamma
    visited = []

    while state not in env.terminal_states:
        if state in visited and len(visited) > 100:
            break
        visited.append(state)

        policy_action = policy[state]
        next_state, reward = env.step(state, policy_action)
        discounted_return += discount * reward
        discount *= gamma
        state = next_state

    return discounted_return

# Display values
s0_value = state_value_function('S0', policy, env, gamma)
s0_action_value_up = action_value_function('S0', 'Up', policy, env, gamma)

print(f"State Value for state 'S0': {s0_value:.4f}")
print(f"Action Value for state 'S0' and action 'Up': {s0_action_value_up:.4f}")
```
