```python
# Assignment 1
# State Value and Action Value for Grid World
# No Bellman Equation is used.

# Load the MDP module from Assignment 7
# import grid_env_det


# ---------------------------------------------------------
# Policy
# ---------------------------------------------------------
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

gamma = 0.9


# ---------------------------------------------------------
# Environment information
# ---------------------------------------------------------
# State transitions according to the given Grid World
# and rewards received after taking an action.

transitions = {
    ("S0", "Right"): ("S1", -1),
    ("S1", "Down"):  ("S3", -1),
    ("S3", "Right"): ("S4", -1),
    ("S4", "Up"):    ("G", 10),

    # Other useful transitions
    ("S0", "Down"):  ("S0", -1),   # obstacle X
    ("S0", "Up"):    ("S0", -1),   # boundary
    ("S0", "Left"):  ("S0", -1),   # boundary

    ("S1", "Up"):    ("S1", -1),
    ("S1", "Left"):  ("S0", -1),
    ("S1", "Right"): ("G", 10),

    ("S3", "Up"):    ("S1", -1),
    ("S3", "Down"):  ("S3", -1),   # boundary
    ("S3", "Left"):  ("S3", -1),

    ("S4", "Down"):  ("S4", -1),
    ("S4", "Right"): ("S4", -1),
    ("S4", "Left"):  ("S3", -1),

    ("G", "Up"):     ("G", 0),
    ("G", "Down"):   ("G", 0),
    ("G", "Left"):   ("G", 0),
    ("G", "Right"):  ("G", 0)
}


# ---------------------------------------------------------
# State Value Function
# ---------------------------------------------------------
def state_value_function(start_state):
    """
    Calculate V(s) by following the given policy
    and calculating the discounted return.

    No Bellman equation is used.
    """

    state = start_state
    total_return = 0
    discount = 1

    while state != "G":

        # Select action according to policy
        action = policy[state]

        # Perform transition
        next_state, reward = transitions[(state, action)]

        # Add discounted reward
        total_return += discount * reward

        # Update discount factor
        discount *= gamma

        # Move to next state
        state = next_state

    return total_return


# ---------------------------------------------------------
# Action Value Function
# ---------------------------------------------------------
def action_value_function(state, action):
    """
    Calculate Q(s,a).

    The first action is forced to be 'action'.
    After that, the given policy is followed.

    No Bellman equation is used.
    """

    total_return = 0
    discount = 1

    # Take the requested first action
    next_state, reward = transitions[(state, action)]

    total_return += discount * reward
    discount *= gamma

    state = next_state

    # Follow the policy after the first action
    while state != "G":

        action = policy[state]

        next_state, reward = transitions[(state, action)]

        total_return += discount * reward
        discount *= gamma

        state = next_state

    return total_return


# ---------------------------------------------------------
# Display results
# ---------------------------------------------------------

# State Value of S0
V_S0 = state_value_function("S0")

# Action Value of S0 for action Up
Q_S0_Up = action_value_function("S0", "Up")

print("State Value Function")
print("--------------------")
print("V(S0) =", V_S0)

print("\nAction Value Function")
print("--------------------")
print("Q(S0, Up) =", Q_S0_Up)
```
