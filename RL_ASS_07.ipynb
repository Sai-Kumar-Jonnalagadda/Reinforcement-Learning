## Assignment 1: Calculate Undiscounted and Discounted Returns

```python
# Assignment 1: Calculate Undiscounted and Discounted Returns

# Accept number of time steps
n = int(input("Enter the number of time steps: "))

# Accept reward values
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Accept discount factor
gamma = float(input("Enter the discount factor (gamma): "))

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Calculate discounted return
discounted_return = 0

for t in range(n):
    discounted_return += (gamma ** t) * rewards[t]

# Display results
print("\n----- RESULTS -----")
print("Entered reward values:", rewards)
print("Undiscounted Return:", undiscounted_return)
print("Discounted Return:", discounted_return)
```

## Assignment 2: 2x3 Grid World as an MDP

grid_env_det.py


```python
%%writefile grid_env_det.py

# Complete Deterministic MDP for 2 x 3 Grid World

# ============================================================
# 1. STATE SPACE
# ============================================================

states = ["S0", "S1", "S3", "S4", "G"]


# ============================================================
# 2. ACTION SPACE
# ============================================================

actions = ["up", "down", "left", "right"]


# ============================================================
# 3. REWARD VALUES
# ============================================================

NORMAL_REWARD = -1
GOAL_REWARD = 10
OBSTACLE_REWARD = -5


# ============================================================
# 4. DISCOUNT FACTOR
# ============================================================

gamma = 0.9


# ============================================================
# 5. TRANSITION FUNCTION
# ============================================================

transition = {

    "S0": {
        "up": "S0",
        "down": "S0",
        "left": "S0",
        "right": "S1"
    },

    "S1": {
        "up": "S1",
        "down": "S3",
        "left": "S0",
        "right": "G"
    },

    "S3": {
        "up": "S1",
        "down": "S3",
        "left": "S3",
        "right": "S4"
    },

    "S4": {
        "up": "G",
        "down": "S4",
        "left": "S3",
        "right": "S4"
    },

    "G": {
        "up": "G",
        "down": "G",
        "left": "G",
        "right": "G"
    }
}


# ============================================================
# 6. REWARD FUNCTION
# ============================================================

def get_reward(current_state, action, next_state):

    if next_state == "G":
        return GOAL_REWARD

    return NORMAL_REWARD


# ============================================================
# 7. NEXT STATE FUNCTION
# ============================================================

def get_next_state(state, action):

    if state not in states:
        raise ValueError("Invalid state!")

    if action not in actions:
        raise ValueError("Invalid action!")

    return transition[state][action]
```

Assignment 2 Main Program

```python
# Assignment 2: 2 x 3 Grid World MDP

import grid_env_det as env


# ============================================================
# DISPLAY STATE SPACE
# ============================================================

print("==========================================")
print("          2 x 3 GRID WORLD MDP")
print("==========================================")

print("\nState Space:")
print(env.states)


# ============================================================
# DISPLAY ACTION SPACE
# ============================================================

print("\nAction Space:")
print(env.actions)


# ============================================================
# DISPLAY REWARD FUNCTION
# ============================================================

print("\nReward Function:")
print("Normal Movement Reward :", env.NORMAL_REWARD)
print("Goal Reward            :", env.GOAL_REWARD)
print("Obstacle Reward        :", env.OBSTACLE_REWARD)


# ============================================================
# DISPLAY DISCOUNT FACTOR
# ============================================================

print("\nDiscount Factor:")
print(env.gamma)


# ============================================================
# DISPLAY TRANSITION FUNCTION
# ============================================================

print("\n==========================================")
print("          TRANSITION FUNCTION")
print("==========================================")

for state in env.states:

    print("\nCurrent State:", state)

    for action in env.actions:

        next_state = env.get_next_state(state, action)

        print(action, "->", next_state)


# ============================================================
# NEXT STATE ESTIMATION
# ============================================================

print("\n==========================================")
print("          NEXT STATE ESTIMATION")
print("==========================================")

current_state = input(
    "Enter current state (S0/S1/S3/S4/G): "
)

action = input(
    "Enter action (up/down/left/right): "
).lower()


# ============================================================
# CALCULATE NEXT STATE AND REWARD
# ============================================================

try:

    next_state = env.get_next_state(
        current_state,
        action
    )

    reward = env.get_reward(
        current_state,
        action,
        next_state
    )

    print("\n========== RESULT ==========")

    print("Current State :", current_state)
    print("Action        :", action)
    print("Next State    :", next_state)
    print("Reward        :", reward)


except ValueError as e:

    print("\nError:", e)
```
