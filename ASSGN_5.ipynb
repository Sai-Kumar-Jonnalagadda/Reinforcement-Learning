## Random Agent for Reinforcement Learning Environments

```python
import gymnasium as gym
import random
import numpy as np

def run_random_agent(env_name, num_episodes=100):
    # 1. Create the selected environment.
    env = gym.make(env_name)
    print(f"Environment created: {env_name}")
    print(f"Observation Space: {env.observation_space}")
    print(f"Action Space: {env.action_space}")

    total_rewards = []

    for episode in range(num_episodes):
        # 2. Reset the environment and 3. Observe the initial state.
        observation, info = env.reset()
        episode_reward = 0
        terminated = False
        truncated = False

        print(f"\n--- Episode {episode + 1} ---")
        # print(f"Initial Observation: {observation}")

        while not terminated and not truncated:
            # 4. Identify the available actions using the environment's action space.
            # 5. Select an action randomly.
            action = env.action_space.sample()
            # print(f"Selected Action: {action}")

            # 6. Execute the action in the environment.
            # 7. Observe the next state and reward.
            observation, reward, terminated, truncated, info = env.step(action)
            episode_reward += reward
            # print(f"Next State: {observation}, Reward: {reward}, Terminated: {terminated}, Truncated: {truncated}")

        total_rewards.append(episode_reward)
        print(f"Episode {episode + 1} finished with reward: {episode_reward}")

    env.close()

    # 9. Calculate and display the total reward and average reward
    #    obtained by the random agent.
    print(f"\n--- Experiment Summary for {env_name} ---")
    print(f"Total episodes: {num_episodes}")
    print(f"Total accumulated reward across all episodes: {np.sum(total_rewards):.2f}")
    print(f"Average reward per episode: {np.mean(total_rewards):.2f}")
    print(f"Standard deviation of reward per episode: {np.std(total_rewards):.2f}")

    return total_rewards
```

### Run for CartPole environment

```python
cartpole_rewards = run_random_agent('CartPole-v1', num_episodes=20)
```

### Run for MountainCar environment

```python
mountaincar_rewards = run_random_agent('MountainCar-v0', num_episodes=20)
```

### Run for FrozenLake environment

```python
# For FrozenLake, we often want to make it non-slippery for easier learning initially
# But for a random agent, the slippery version is fine to demonstrate randomness.
# If you want to try the non-slippery version, use: gym.make('FrozenLake-v1', is_slippery=False)
frozenlake_rewards = run_random_agent('FrozenLake-v1', num_episodes=100)
```
